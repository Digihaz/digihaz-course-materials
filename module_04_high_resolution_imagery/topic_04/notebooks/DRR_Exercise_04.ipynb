```python
## Create ortho
d_groundel = 1
ortho = np.zeros(
    [height_ortho // d_groundel, width_ortho // d_groundel, 3]
    )

print(height_ortho, width_ortho)

for i in tqdm(range(height_ortho // d_groundel)):
    for j in range(width_ortho // d_groundel):
        # Correspondence of DSM coordinates to 3D space coordinates Χ,Υ,Z.
        X = orX_ortho + (j-1)*d_ortho*d_groundel
        Y = orY_ortho - (i-1)*d_ortho*d_groundel
        # elevation interpolation
        # TODO Pick up "Z" using the value of point (i,j) in matrix 'dsm'.
        # Z = ....
        
        # Project Χ,Υ,Z (DSM-based coordinates) to x,y (projected image plane)
        # Use the collinearity condition
        #U = ...
        #V = ...
        #W = ...
        x = xo - c*U/W
        y = yo - c*V/W
       
        # convert x,y- coordinates to i,j- coordinates
        im_j = x/d_im + width/2
        im_i = height/2 - y/d_im
        # skip points outside image frame        
        if(im_i<0 or im_j<0 or im_i>=height-1 or im_j>=width-1):
            continue
        # image interpolation
        im_j = np.uint16(np.round(im_j))
        im_i = np.uint16(np.round(im_i))
        #ortho[i, j, :] = im[im_i, im_j, :]
        ortho[i, j, :] = im[im_i, im_j, :]

```
